```python
# Step 1 Data Storage

employees = {
    101: {'Name': 'Satya', 'Age': 27, 'Department': 'HR', 'Salary': 50000}
}
```

```python
# step 2 Add Menu Syste,

def main_menu():
    for i in range(100):   
        print("\nMenu")
        print("1. Add Employee")
        print("2. View all Employees")
        print("3. Search for Employee")
        print("4. Exit")

        action = int(input("\nEnter the number of Action (1 to 4): "))

        if action == 1:
            print("Add Employee")
            add_employee()
        elif action == 2:
            print("View all Employees")
            view_all()
        elif action == 3:
            print("Search an Employee")
            search(emp_id)
        elif action == 4:
            print("Exit")
            break
        else:
            print("Enter valid action number")
```

```python
# Step 3 Add Employee Functionality

def add_employee():
    emp_id = int(input("Enter Employee's ID:"))
    
    if emp_id in employees:
        print("Employee already exists!")
    else:
        name = input("Enter Employee's Name:")
        age = int(input("Enter Employee's Age:"))
        department = input("Enter Employee's Department:")
        salary = int(input("Enter Employee's Salary:"))
        
        employees[emp_id] = {"Name": name, "Age": age, "Department": department, "Salary": salary}
        print("Employee added successfully!")
```

```python
# Step 4 View all employees

def view_all():
    if not employees:
        print("Employee not found!")
    else:
        for emp_id in employees:
            print(f"{emp_id}\t{employees[emp_id]['Name']}\t{employees[emp_id]['Age']}\t{employees[emp_id]['Department']}\t{employees[emp_id]['Salary']}")
```

```python
# Step 5 search an employee by emp_id

def search():
    emp_id = int(input("Enter Employee's ID to search:"))
    
    if emp_id in employees:
        print(f"\nName: {employees[emp_id]['Name']}")
        print(f"Age: {employees[emp_id]['Age']}")
        print(f"Department: {employees[emp_id]['Department']}")
        print(f"Salary: Rs {employees[emp_id]['Salary']}")
    else:
        print("Employee not found!")
```

```python
main_menu()
```
